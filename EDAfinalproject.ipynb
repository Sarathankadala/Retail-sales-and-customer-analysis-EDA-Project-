# **Derived Columns**

**1. Create GrossAmount = Quantity × UnitPrice**

```python
df["GrossAmount"] = df["Quantity"] * df["UnitPrice"]
```

**2. Create DiscountAmount = GrossAmount × DiscountPct / 100.**

```python
df["DiscountAmount"] = df["GrossAmount"] * df["DiscountPct"] / 100
```

**3. Create NetSales = GrossAmount - DiscountAmount**

```python
df["NetSales"] = df["GrossAmount"] - df["DiscountAmount"]
```

**4. Use apply() to classify NetSales as High Value (> 50,000) or Regular**

```python
def salescal(sales):
    if sales > 50000:
        return "High Value"
    else:
        return "Regular"

df["SalesType"] = df["NetSales"].apply(salescal)
```

**5. Create OrderMonth from OrderDate**

```python
df["OrderMonth"] = df["OrderDate"].dt.to_period("M")
```
